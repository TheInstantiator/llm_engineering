## The Chinchilla Scaling Law

Number of parameters is proportional to the amount of training tokens.

If you are getting diminishing returns from training with more training data, you should be adding more parameters to the model.
Also if you upgrade the model to double the number of parameters you should double the amount of training data.

## Understanding Benchmarks ( 6 hard benchmarks )

- GPQA - Google Proof Q and A
  - 448 questions if you give to non PHD level people with google search as a reference they would score 34%
    - PHD Level is 65%
  - MMLU-PRO (Massive multi-modal language understanding)
  - AIME a competative exam for high school students in math
  - LiveCode Bench hard leetcode like problems
  - MuSR - reasoning, it is giving 1000 word murder mysteryt and asked who has the means, motive, opportunity
  - HLE - Humanities last exam, 2500 of the toughest subject diverse multi modal questions

## Limitations of Benchmarks
- Training data contamination : questions get into the training data
- Not consistently applied
- narrow scope like only in math
- hard to measure nuanced reasoning
- Saturation - models are getting better at the same rate
- Overfitting to benchmarks
- Frontier models may be aware that they are being benchmarked or not
