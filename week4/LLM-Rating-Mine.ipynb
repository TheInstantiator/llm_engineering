# 🎯 What is the Right LLM for the Job?

Choosing the right Large Language Model isn't just about picking the "best" one—it's about finding the **right fit** for your specific use case.

---

## 📋 The Basics

### Key Features to Consider
| Feature | What to Look For |
|---------|-----------------|
| **Open vs Closed Source** | Trade-off between flexibility and support |
| **Model Type** | Chat / Reasoning / Hybrid capabilities |
| **Knowledge Cutoff** | How recent is the training data? |

### Critical Metrics

- **Parameters** — The number of parameters the model was trained on *(bigger ≠ always better!)*
- **Context Length** — How many tokens the model can "remember" in a conversation
- **Inference Speed** — How fast can it generate responses?
- **Price** — The *real* cost breakdown:
  - 💰 API cost *or* runtime compute cost
  - 🏋️ Training cost
  - 🔧 Build cost
  - ⏱️ Time to market for your product
  - ⚡ Rate limiting / latency / licensing restrictions

---

## 📊 Performance: How Do We Measure It?

Three key ways to evaluate LLM performance:
1. **Benchmarks** — Standardized tests
2. **Leaderboards** — Comparative rankings
3. **Arenas** — Head-to-head competitions

---

## 🐿️ The Chinchilla Scaling Law

> *"Number of parameters is proportional to the amount of training tokens."*

**Key Insights:**
- Getting **diminishing returns** from more training data? → *Add more parameters*
- **Doubling parameters?** → *Double your training data too*

This law fundamentally changed how we think about scaling LLMs efficiently.

## 🧪 Understanding Benchmarks (6 Hard Benchmarks)

| Benchmark | Description | Why It Matters |
|-----------|-------------|----------------|
| **GPQA** | Google-Proof Q&A — 448 questions where non-PhD scorers with Google only get **34%**, PhD-level gets **65%** | Tests true reasoning, not search skills |
| **MMLU-PRO** | Massive Multi-modal Language Understanding | Broad knowledge assessment |
| **AIME** | Competitive math exam for high schoolers | Mathematical reasoning depth |
| **LiveCodeBench** | Hard LeetCode-style problems | Real coding ability |
| **MuSR** | 1000-word murder mysteries: *Who has means, motive, opportunity?* | Complex reasoning chains |
| **HLE** | Humanity's Last Exam — 2500 toughest multi-modal questions | The ultimate stress test |

---

## ⚠️ Limitations of Benchmarks

Benchmarks aren't perfect. Watch out for:

- 🔴 **Training Data Contamination** — Questions leak into training data
- 🔴 **Inconsistent Application** — Not applied the same way across models
- 🔴 **Narrow Scope** — e.g., only testing math
- 🔴 **Hard to Measure Nuanced Reasoning** — Subtlety gets lost
- 🔴 **Saturation** — All models improving at similar rates
- 🔴 **Overfitting to Benchmarks** — Optimizing for tests, not real-world use
- 🔴 **Benchmark Awareness** — Frontier models may *know* they're being tested 👀

# 💼 Defining the Business Problem

> **Everyone says:** *"I want an AI Agent!"*

But that's not enough. Before building anything, you need to dig deeper.

---

## 🔍 The Critical Questions

Before ANY AI project, ask:

| Question | Why It Matters |
|----------|---------------|
| **What business problem are you solving?** | No clear problem = no clear solution |
| **What is the value to the customer/company/yourself?** | Quantify the impact |
| **What is the expected outcome?** | Define success upfront |
| **How do you measure success?** | If you can't measure it, you can't improve it |

---

**Them:** *"I want to be able to ask: what are my sales numbers today?"*

> "Okay, but **what business problem** are you actually trying to solve?
> - What's the **value** of having this instantly?

> - How will you **measure success**?"

# 📍 The 5-Step Strategy for LLM Selection

A systematic approach to choosing and deploying the right model:

---

```
┌─────────────────────────────────────────────────────────────┐
│  1️⃣  UNDERSTAND  →  2️⃣  PREPARE  →  3️⃣  SELECT  →  4️⃣  ITERATE  →  5️⃣  DEPLOY  │
└─────────────────────────────────────────────────────────────┘
```

---

| Step | Action | Key Activities |
|------|--------|----------------|
| **1️⃣ Understand** | Define the problem | Clarify requirements, constraints, success metrics |
| **2️⃣ Prepare** | Select candidate models | Research options, compare benchmarks, check costs |
| **3️⃣ Select** | Build & prototype | Create quick POCs with top candidates |
| **4️⃣ Iterate** | Refine the model | Prompt engineering, fine-tuning, evaluation loops |
| **5️⃣ Deploy** | Go to production | Monitoring, scaling, continuous improvement |

---

> 💡 **Pro Tip:** Don't skip step 1! A well-understood problem is half-solved.

# How to evaluate the performance of an AI Solution

# Types of Evaluation

## Model-centric or Technical Evaluation

- loss
  - MSE - mean squared error
    - Take the difference between the predicted value and the actual value, square it, and then take the average of all the squared differences.
  - MAE - mean absolute error
      - Take the absolute value of the difference between the predicted value and the actual value, and then take the average of all the absolute differences.
  - RMSE - root mean squared error
    - Take the square root of the mean squared error.

- perplexity
  - 1 is confident, higher is less confident

- accuracy

- precision

- recall

- F1 score

- ROC AUC

## User-centric or Business Evaluation

- KPIs
- ROI
- Customer satisfaction  

Business evaluation can be more fuzzy and less objective.
Why are customers not satisfied with the AI solution?  It may not have to do with the AI solution at all.
Why is ROI low?  It may not have to do with the AI solution at all.  

So generally focus on model-centric evaluation first, and then business evaluation.

As an engineer you mainly care about model-centric evaluation but to be a good engineer you should also understand business evaluation.
