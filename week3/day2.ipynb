Higher level APIs to carry out standard tasks incredibly quickly

Best for simple out-of-the-box inference taks.<br>The let you unleash the power of open-source models in your solutions in 2 lines of code!<br>
Example: Sentiment analysis, classifiers, named entity recognition, question anwering, summarizing, and translation<br>
