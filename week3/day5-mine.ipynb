### Using other models

The visualizer can be used with any model that provides an OpenAI-compatible API. This includes services like Groq and local models served with Ollama.

#### Groq

To use Groq, you will need to get a Groq API key and set it as an environment variable. You can name the environment variable `GROQ_API_KEY` and the .env file will load it. Then, you can use the visualizer by specifying the Groq model and the Groq API endpoint.

#### Gemini

The Gemini API is not compatible with the OpenAI SDK. To use Gemini, you would need to create a new `TokenPredictor` class that uses the `google-generativeai` library. This would involve:
1.  Installing the `google-generativeai` library: `pip install google-generativeai`
2.  Creating a new class (e.g., `GeminiTokenPredictor`) that uses `genai.GenerativeModel` to generate content.
3.  The Gemini API does not currently support streaming with logprobs, so you would need to find a way to get token probabilities. This might require using a different method or a different model.
