## First - let's talk about the Chat Completions API

1. The simplest way to call an LLM
2. It's called Chat Completions because it's saying: "here is a conversation, please predict what should come next"
3. The Chat Completions API was invented by OpenAI, but it's so popular that everybody uses it!

### We will start by calling OpenAI again - but don't worry non-OpenAI people, your time is coming!


````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)

api_key = os.getenv('OPENAI_API_KEY')
grok_api_key = os.getenv('GROK_API_KEY')
gemini_api_key = os.getenv('GEMINI_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")

if not grok_api_key:
    print("No Grok API key was found - please set GROK_API_KEY in your .env file if you want to use Grok models.")
elif not grok_api_key.startswith("xai-"):
    print("A Grok API key was found, but it doesn't start sk-grok-; please check you're using the right key - see troubleshooting notebook")
else:
    print("Grok API key found and looks good so far!")

if not gemini_api_key:
    print("No Gemini API key was found - please set GEMINI_API_KEY in your .env file if you want to use Gemini models.")
elif not gemini_api_key.startswith("AIz"):
    print("A Gemini API key was found, but it doesn't start sk-gemini-; please check you're using the right key - see troubleshooting notebook")
else:
    print("Gemini API key found and looks good so far!")
````

## Do you know what an Endpoint is?

If not, please review the Technical Foundations guide in the guides folder

And, here is an endpoint that might interest you...

````python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}
grok_headers = {"Authorization": f"Bearer {grok_api_key}", "Content-Type": "application/json"}
gemini_headers = {"Authorization": f"Bearer {gemini_api_key}", "Content-Type": "application/json"}

BASE_URL = "https://api.openai.com/v1/chat/completions"
GROK_BASE_URL = "https://api.x.ai/v1"
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai"

payload = {
    "model": "gpt-5-nano",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

grok_payload =  {
    "model": "grok-4",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

gemini_payload =  {
    "model": "gemini-2.5-pro",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}
payload
````

````python
response = requests.post(
    BASE_URL,
    headers=headers,
    json=payload
)

response.json()
````

````python
response.json()["choices"][0]["message"]["content"]
````

````python
response_grok = requests.post(
    f"{GROK_BASE_URL}/chat/completions",
    headers=grok_headers,
    json=grok_payload
)

response_grok.json()
````

# What is the openai package?

It's known as a Python Client Library.

It's nothing more than a wrapper around making this exact call to the http endpoint.

It just allows you to work with nice Python code instead of messing around with janky json objects.

But that's it. It's open-source and lightweight. Some people think it contains OpenAI model code - it doesn't!


````python
response_grok.json()["choices"][0]["message"]["content"]
````

````python
response_gemini = requests.post(
    f"{GEMINI_BASE_URL}/chat/completions",
    headers=gemini_headers,
    json=gemini_payload
)

response_gemini.json()
````

````python
# Create OpenAI client

openai = OpenAI()

response = openai.chat.completions.create(model="gpt-5-nano", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content

````

## And then this great thing happened:

OpenAI's Chat Completions API was so popular, that the other model providers created endpoints that are identical.

They are known as the "OpenAI Compatible Endpoints".

For example, google made one here: https://generativelanguage.googleapis.com/v1beta/openai/

And OpenAI decided to be kind: they said, hey, you can just use the same client library that we made for GPT. We'll allow you to specify a different endpoint URL and a different key, to use another provider.

So you can use:

```python
gemini = OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/", api_key="AIz....")
gemini.chat.completions.create(...)
```

And to be clear - even though OpenAI is in the code, we're only using this lightweight python client library to call the endpoint - there's no OpenAI model involved here.

If you're confused, please review Guide 9 in the Guides folder!

And now let's try it!

````python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not google_api_key.startswith("AIz"):
    print("An API key was found, but it doesn't start AIz")
else:
    print("API key found and looks good so far!")

````

````python
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=gemini_api_key)

response = gemini.chat.completions.create(model="gemini-2.5-pro", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content
````

````python
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI

openai = OpenAI()

OLLAMA_BASE_URL = "http://localhost:11434/v1"

# api key can be anything, it just needs to be there
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
````

````python
# System prompt - defines the AI's role and output format
system_prompt = """
You are an NFL injury report analyst. Your job is to:
1. Extract player injury information from website content
2. Format it in a clear, structured way
3. Include only relevant injury/status information

For each player, provide:
- Name
- position
- Injury type
- Practice participation status for Wed, Thu, Fri.  Use the legeend at the bottom of the report to interpret the abbreviations.  Don't display the abbreviations, only the interpretation in full.
- Game status (Out/Questionable/Doubtful/etc.)

If a column does not have any data for a player, display - in the cell
Format the response in markdown tables for readability.
"""

# User prompt - gives the specific task and any special instructions
user_prompt_prefix = """
Analyze this NFL injury report webpage and extract all player injury information.
Only include actual injury report data, ignore navigation menus and other website elements.
Present the information in a table format with columns for Name, Injury, Practice Status, and Game Status.

Website content to analyze:
"""

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
def summarize(url, model):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
````

````python
url = "https://www.dallascowboys.com/team/injury-report/"
````

````python
summarizey = summarize(url, model="phi4")
display(Markdown(summarizey))
````

````python
summary = summarize(url, model="gemma3")
display(Markdown(summary))
````

````python
summary = summarize(url, model="gpt-oss")
display(Markdown(summary))
````

````python
summary = summarize(url, model="deepseek-r1")
display(Markdown(summary))
````

````python
summary = summarize(url, model="llama3.2")
display(Markdown(summary))
````
